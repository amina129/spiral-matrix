```python
import time
import os
import sys

def clear_screen():
    os.system('cls' if os.name == 'nt' else 'clear')

def spiral_traverse_with_visualization(matrix, delay=0.5):

    if not matrix or not matrix[0]:
        return []

    result = []
    n = len(matrix)

    # Create a copy for visualization
    visual_matrix = [row[:] for row in matrix]

    # Initialize boundaries
    top, bottom = 0, n - 1
    left, right = 0, n - 1

    step_count = 0
    total_steps = n * n

    while top <= bottom and left <= right:
        # Traverse from left to right along the top row
        for col in range(left, right + 1):
            clear_screen()
            print("=" * 50)
            print(f"SNAIL MATRIX TRAVERSAL - Step {step_count + 1}/{total_steps}")
            print("=" * 50)

            # Show the current element being visited
            current_value = matrix[top][col]
            result.append(current_value)

            # Create visualization
            for i in range(n):
                for j in range(n):
                    if i == top and j == col:
                        # Current position
                        print(f"\033[92m▶ {visual_matrix[i][j]:3}\033[0m", end=" ")
                    elif (i < top or i > bottom or j < left or j > right):
                        # Already traversed (inner ring)
                        print(f"  {visual_matrix[i][j]:3}", end=" ")
                    elif (i == top and j < col) or (i == bottom and j > col) or \
                         (j == right and i < top) or (j == left and i > bottom):
                        # Already traversed in current ring
                        print(f"\033[93m  {visual_matrix[i][j]:3}\033[0m", end=" ")
                    else:
                        # Not yet traversed
                        print(f"  {visual_matrix[i][j]:3}", end=" ")
                print()

            print("\n\033[92m▶\033[0m Currently visiting: ", current_value)
            print("Spiral order so far: ", " → ".join(str(x) for x in result))
            print(f"\nTraversing: Top row → ({top}, {col})")

            # Mark as visited in visual matrix
            visual_matrix[top][col] = f"\033[90m{visual_matrix[top][col]}\033[0m"

            step_count += 1
            time.sleep(delay)

        top += 1

        if top > bottom:
            break

        # Traverse from top to bottom along the right column
        for row in range(top, bottom + 1):
            clear_screen()
            print("=" * 50)
            print(f"SNAIL MATRIX TRAVERSAL - Step {step_count + 1}/{total_steps}")
            print("=" * 50)

            current_value = matrix[row][right]
            result.append(current_value)

            for i in range(n):
                for j in range(n):
                    if i == row and j == right:
                        print(f"\033[92m▶ {matrix[i][j]:3}\033[0m", end=" ")
                    elif (i < top-1 or i > bottom or j < left or j > right):
                        print(f"  {matrix[i][j]:3}", end=" ")
                    elif (i == top-1 and j <= right) or (i == bottom and j >= left) or \
                         (j == right and i <= row) or (j == left and i >= top-1):
                        print(f"\033[93m  {matrix[i][j]:3}\033[0m", end=" ")
                    else:
                        print(f"  {matrix[i][j]:3}", end=" ")
                print()

            print("\n\033[92m▶\033[0m Currently visiting: ", current_value)
            print("Spiral order so far: ", " → ".join(str(x) for x in result))
            print(f"\nTraversing: Right column ↓ ({row}, {right})")

            step_count += 1
            time.sleep(delay)

        right -= 1

        if left > right:
            break

        # Traverse from right to left along the bottom row
        for col in range(right, left - 1, -1):
            clear_screen()
            print("=" * 50)
            print(f"SNAIL MATRIX TRAVERSAL - Step {step_count + 1}/{total_steps}")
            print("=" * 50)

            current_value = matrix[bottom][col]
            result.append(current_value)

            for i in range(n):
                for j in range(n):
                    if i == bottom and j == col:
                        print(f"\033[92m▶ {matrix[i][j]:3}\033[0m", end=" ")
                    elif (i < top or i > bottom or j < left or j > right+1):
                        print(f"  {matrix[i][j]:3}", end=" ")
                    elif (i == bottom and j >= col) or (i == top-1 and j <= right+1) or \
                         (j == left and i >= top) or (j == right+1 and i <= bottom):
                        print(f"\033[93m  {matrix[i][j]:3}\033[0m", end=" ")
                    else:
                        print(f"  {matrix[i][j]:3}", end=" ")
                print()

            print("\n\033[92m▶\033[0m Currently visiting: ", current_value)
            print("Spiral order so far: ", " → ".join(str(x) for x in result))
            print(f"\nTraversing: Bottom row ← ({bottom}, {col})")

            step_count += 1
            time.sleep(delay)

        bottom -= 1

        if top > bottom:
            break

        # Traverse from bottom to top along the left column
        for row in range(bottom, top - 1, -1):
            clear_screen()
            print("=" * 50)
            print(f"SNAIL MATRIX TRAVERSAL - Step {step_count + 1}/{total_steps}")
            print("=" * 50)

            current_value = matrix[row][left]
            result.append(current_value)

            for i in range(n):
                for j in range(n):
                    if i == row and j == left:
                        print(f"\033[92m▶ {matrix[i][j]:3}\033[0m", end=" ")
                    elif (i < top or i > bottom+1 or j < left or j > right):
                        print(f"  {matrix[i][j]:3}", end=" ")
                    elif (i == bottom+1 and j <= right) or (i == top and j >= left) or \
                         (j == left and i <= row) or (j == right and i >= top):
                        print(f"\033[93m  {matrix[i][j]:3}\033[0m", end=" ")
                    else:
                        print(f"  {matrix[i][j]:3}", end=" ")
                print()

            print("\n\033[92m▶\033[0m Currently visiting: ", current_value)
            print("Spiral order so far: ", " → ".join(str(x) for x in result))
            print(f"\nTraversing: Left column ↑ ({row}, {left})")

            step_count += 1
            time.sleep(delay)

        left += 1

    # Final display
    clear_screen()
    print("=" * 50)
    print("TRAVERSAL COMPLETE!")
    print("=" * 50)
    print("\nFinal Spiral Order:")
    print(" → ".join(str(x) for x in result))
    print(f"\nTotal elements: {len(result)}")

    return result

matrix = [
    [1,   8,  3,  4,  5],
    [76, 13, 18, 19,  6],
    [15, 94, 95, 91,  7],
    [14, 23, 29, 71,  8],
    [13, 27, 11, 10,  9]
]

def print_matrix(matrix):
    """Helper function to print the matrix in a readable format"""
    print("Matrix:")
    for row in matrix:
        print("  ".join(f"{num:3}" for num in row))
    print()

if __name__ == "__main__":
    print_matrix(matrix)
    input("Press Enter to start the spiral visualization...")

    # Run the spiral traversal with visualization
    result = spiral_traverse_with_visualization(matrix, delay=0.8)

    print("\n\nFinal result:", result)
```
